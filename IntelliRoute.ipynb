## Data preprocessing

Since this data set contains the column 1 being the message and column 2 being the department it can be viewed as:

            messages (X) -> department (Y)

Menaing that this is a classification task. Due to the semantic nature of the messages GPT2 is used for understanding the mapping between the message and the department

```python
# split the data set into training, validation and testing
from sklearn.model_selection import train_test_split
import pandas as pd

df = pd.read_csv("dataset/Customer-Responses.csv")

# 70% of the data set will be for training. Basically 70% goes into train_df and the 30% goes into temp_df 
train_df, temp_df = train_test_split(
    df,
    test_size=0.3,
    random_state=42,
    stratify=df["department"]
)

# the rest 30% gets distributed into validation and testing
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.5,
    random_state=42,
    stratify=temp_df["department"]
)

print(len(train_df))
print(len(val_df))
print(len(test_df))
print(len(df))
```

```python
# seperate messages from departments
train_messages = train_df["message"].tolist()
val_messages = val_df["message"].tolist()
test_messages = test_df["message"].tolist()

# convert the departments into numberical labels
label_mapping = {
    "billing": 0,
    "returns": 1,
    "general": 2,
    "pharmacy": 3
}

train_labels = train_df["department"].map(label_mapping).tolist()
val_labels = val_df["department"].map(label_mapping).tolist()
test_labels = test_df["department"].map(label_mapping).tolist()

print(f"Size of train messages: {len(train_messages)}. Size of train labels: {len(train_labels)}")
print(f"Size of validation messages: {len(val_messages)}. Size of validation labels: {len(val_labels)}")
print(f"Size of test messages: {len(test_messages)}. Size of test labels: {len(test_labels)}")
```

```python
# tokenize the messages from train, validate and test
import tiktoken

tokenizer = tiktoken.get_encoding("gpt2")

# tokenize each training message
train_tokens = [
    tokenizer.encode(message) for message in train_messages
]

# tokenize each training message
validation_tokens = [
    tokenizer.encode(message) for message in val_messages
]

# tokenize each training message
test_tokens = [
    tokenizer.encode(message) for message in test_messages
]

print(f"Size of train tokens: {len(train_tokens)}")
print(f"Size of train tokens: {len(validation_tokens)}")
print(f"Size of train tokens: {len(test_tokens)}")
```

## Dataset and dataloader

```python
# preparing tokenized messages so that pytorch can feed into GPT during training
import torch
from torch.utils.data import Dataset, DataLoader

class CustomerDataset(Dataset):
    def __init__(self, tokens, labels, max_length):
        self.labels = labels
        self.max_length = max_length
        self.pad_token_id = 50256

        # store customer messages after truncation and padding
        self.tokens = []
        # number of real tokens in each message after truncation but before padding
        self.lengths = []

        # padding. Making all of the wods the same length. If not then 50256 is added at the end to make all max_length
        for token_ids in tokens:
            # turncate if needed
            token_ids = token_ids[:max_length]
            length = len(token_ids)
            
            # pad to fixed length
            token_ids = token_ids + [self.pad_token_id] * (max_length - length)

            self.tokens.append(token_ids)
            self.lengths.append(length)

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return (
            torch.tensor(self.tokens[idx], dtype=torch.long),
            torch.tensor(self.labels[idx], dtype=torch.long),
            torch.tensor(self.lengths[idx], dtype=torch.long)
        )
```

```python
# initialize datasets
max_length = min(256, max(len(tokens) for tokens in train_tokens))

train_dataset = CustomerDataset(tokens=train_tokens, labels=train_labels, max_length=max_length)
validation_dataset = CustomerDataset(tokens=validation_tokens, labels=val_labels, max_length=max_length)
test_dataset = CustomerDataset(tokens=test_tokens, labels=test_labels, max_length=max_length)

print(f"Length of training set: {len(train_dataset)}")
print(f"Length of validation set: {len(validation_dataset)}")
print(f"Length of test set: {len(test_dataset)}")
```

```python
# datalaoder
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
validate_loader = DataLoader(validation_dataset, batch_size=16, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=16, shuffle=False)
```

## Modifying the output layer

Since our model will not be used to generate text its important to chnage the final layer (output layer). In a langaue generating GPT2 the output layer will go from number of dimenions to vocab size to predict the next token to ouput. 

However, in our case we want the GPT2 to predict which department the message belongs to.

## Partial fine tuning VS Full fine tuning (with training)

After doing some research about fine tuning I understood that there are 2 types of fine tuning that I could do for this project:

1. Partial fine tuning - train the last transformer block + classification head. Its faster and uses less GPU memory. It has lower risk of overfitting but has a lower ability to adapt. The accuracy could be better. 
2. Full fine tuning - train all the 12 blocks + embeddings + head (got GPT2). Its slower and uses more GPU memory. There is a higher overfitting risk however it has a higher ability to adapt. The accuracy is abit better.

```python
import torch
import torch.nn as nn
from transformers import GPT2LMHeadModel
from tqdm.auto import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

def create_model():
    torch.manual_seed(42)
    model = GPT2LMHeadModel.from_pretrained("gpt2")
    model.lm_head = nn.Linear(model.config.n_embd, 4)
    return model.to(device)

def fine_tune(model, learning_rate, epochs=5):
    optimizer = torch.optim.AdamW(
        (p for p in model.parameters() if p.requires_grad),
        lr=learning_rate
    )

    losses = []

    for epoch in range(epochs):
        model.train()
        total_loss = 0

        for inputs, labels, lengths in tqdm(train_loader, desc=f"Epoch {epoch + 1}/{epochs}"):
            inputs = inputs.to(device)
            labels = labels.to(device)
            lengths = lengths.to(device)

            attention_mask = (
                torch.arange(inputs.shape[1], device=device)[None, :] < lengths[:, None]
            ).long()

            optimizer.zero_grad()

            outputs = model(
                input_ids=inputs,
                attention_mask=attention_mask,
                use_cache=False
            )

            batch_idx = torch.arange(inputs.size(0), device=device)
            logits = outputs.logits[batch_idx, lengths - 1]
            loss = nn.functional.cross_entropy(logits, labels)

            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()

            total_loss += loss.item() * labels.size(0)

        avg_loss = total_loss / len(train_dataset)
        losses.append(avg_loss)

    return losses
```

### Partial fine tuning

```python
partial_model = create_model()

# freeze all parameters
for param in partial_model.parameters():
    param.requires_grad = False

# unfreeze final transformer block
for param in partial_model.transformer.h[-1].parameters():
    param.requires_grad = True

# unfreeze final layer normalization
for param in partial_model.transformer.ln_f.parameters():
    param.requires_grad = True

# unfreeze classification head
for param in partial_model.lm_head.parameters():
    param.requires_grad = True

partial_losses = fine_tune(
    partial_model,
    learning_rate=1e-4,
    epochs=5
)

# move model off GPU
partial_model.cpu()
torch.cuda.empty_cache()
```

### Full fine tuning

```python
full_model = create_model()

# unfreeze all parameters
for param in full_model.parameters():
    param.requires_grad = True

full_losses = fine_tune(
    full_model,
    learning_rate=2e-5,
    epochs=5
)
```

## Validation and comparisson

I wanted to see the diffrence between full tuning and partial fine tuning

From the graph its obvious that both fine tuning models are learning. However we can see that the full fine tuning achieves a lower loss than the partial fine tuning.

Now I would want to see validation loss between the two

```python

def evaluate_model(model, dataloader):
    model.to(device)
    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():
        for inputs, labels, lengths in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            lengths = lengths.to(device)

            attention_mask = (
                torch.arange(inputs.shape[1], device=device)[None, :]
                < lengths[:, None]
            ).long()

            outputs = model(
                input_ids=inputs,
                attention_mask=attention_mask,
                use_cache=False
            )

            batch_idx = torch.arange(inputs.size(0), device=device)
            logits = outputs.logits[batch_idx, lengths - 1]

            loss = torch.nn.functional.cross_entropy(logits, labels)

            predictions = torch.argmax(logits, dim=-1)

            total_loss += loss.item() * labels.size(0)
            correct += (predictions == labels).sum().item()
            total += labels.size(0)

    return total_loss / total, correct / total

partial_val_loss, partial_acc = evaluate_model(partial_model, validate_loader)

# Free GPU memory before evaluating the next model
partial_model.cpu()
torch.cuda.empty_cache()

full_val_loss, full_acc = evaluate_model(full_model, validate_loader)

print(f"Partial: Loss={partial_val_loss:.4f}, Accuracy={partial_acc:.2%}")
print(f"Full:    Loss={full_val_loss:.4f}, Accuracy={full_acc:.2%}")
```

## Accuracy

```python
import torch
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    ConfusionMatrixDisplay
)

class_names = ["Billing", "Returns", "General", "Pharmacy"]

def test_model(model, dataloader):
    model.to(device)
    model.eval()

    all_predictions = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels, lengths in dataloader:
            inputs = inputs.to(device)
            lengths = lengths.to(device)

            attention_mask = (
                torch.arange(inputs.shape[1], device=device)[None, :]
                < lengths[:, None]
            ).long()

            outputs = model(
                input_ids=inputs,
                attention_mask=attention_mask,
                use_cache=False
            )

            batch_idx = torch.arange(inputs.size(0), device=device)
            logits = outputs.logits[batch_idx, lengths - 1]

            predictions = torch.argmax(logits, dim=-1)

            all_predictions.extend(predictions.cpu().tolist())
            all_labels.extend(labels.tolist())

    return all_labels, all_predictions
```

```python
# Evaluate partial model
partial_true, partial_pred = test_model(
    partial_model, test_loader
)

partial_model.cpu()

# Evaluate full model
full_true, full_pred = test_model(
    full_model, test_loader
)

print("PARTIAL FINE-TUNING")
print(classification_report(
    partial_true, partial_pred,
    labels=[0, 1, 2, 3],
    target_names=class_names,
    zero_division=0
))

print("\nFULL FINE-TUNING")
print(classification_report(
    full_true, full_pred,
    labels=[0, 1, 2, 3],
    target_names=class_names,
    zero_division=0
))
```

```python
for name, y_true, y_pred in [
    ("Partial Fine-Tuning", partial_true, partial_pred),
    ("Full Fine-Tuning", full_true, full_pred)
]:
    ConfusionMatrixDisplay.from_predictions(
        y_true,
        y_pred,
        labels=[0, 1, 2, 3],
        display_labels=class_names,
        cmap="Blues"
    )

    plt.title(name)
    plt.show()
```

## inccorectly classification

```python
def show_errors(y_true, y_pred, test_df):
    results = test_df[["message"]].reset_index(drop=True).copy()

    results["Actual"] = [
        class_names[i] for i in y_true
    ]

    results["Predicted"] = [
        class_names[i] for i in y_pred
    ]

    errors = results[
        results["Actual"] != results["Predicted"]
    ]

    print(f"Incorrect classifications: {len(errors)}")

    return errors


# Requires test_loader to use shuffle=False
partial_errors = show_errors(partial_true, partial_pred, test_df)
full_errors = show_errors(full_true, full_pred, test_df)

display(partial_errors.head(20))
display(full_errors.head(20))
```
